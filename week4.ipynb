```python
#Data Cleaning & Preprocessing
#Handle missing values
import seaborn as sns
import pandas as pd
df = sns.load_dataset("titanic")
#Fill missing values
df['age'].fillna(df['age'].median(), inplace=True)
df['embarked'].fillna(df['embarked'].mode()[0], inplace=True)
#Drop duplicates
df.drop_duplicates(inplace=True)
#Encode categorial variables
df = pd.get_dummies(df, columns=['sex','class','embarked'], drop_first=True)
#Feature enginnering: family size
df['family_size'] = df['sibsp']+df['parch']
print(df.head())
```

```python
# Data Cleaning & Preprocessing
# Import libraries
import seaborn as sns
import pandas as pd
# Load student.csv
df = pd.read_csv("student.csv")
# Handle missing values
# Fill missing numerical values
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Attendance'].fillna(df['Attendance'].median(), inplace=True)
df['Math_Score'].fillna(df['Math_Score'].median(), inplace=True)
df['Science_Score'].fillna(df['Science_Score'].median(), inplace=True)
# Fill missing categorical values
df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)
df['Department'].fillna(df['Department'].mode()[0], inplace=True)
df['City'].fillna(df['City'].mode()[0], inplace=True)
# Drop duplicates
df.drop_duplicates(inplace=True)
# Encode categorical variables
df = pd.get_dummies(df, columns=['Gender', 'Department', 'City'], drop_first=True)
# Feature Engineering: Average Score
df['Average_Score'] = (df['Math_Score'] + df['Science_Score']) / 2
# Feature Engineering: Total Score
df['Total_Score'] = df['Math_Score'] + df['Science_Score']
# Display the cleaned data
print(df.head())
```
