```python
import pandas as pd
import numpy as np
df=pd.DataFrame({'Age': [25,30,np.nan,40,35],'Department': ['HR','Finance','Finance',np.nan,'IT']})
print(df)
#mean for numeric
df['Age']=df['Age'].fillna(df['Age'].mean())
#mode for categorial
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
print("\nAfter Filling Missing Values:")
print(df)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    'Id': [1,2,2,3,4,4],
    'Name': ['sindhu', 'Chetna', 'Chetna', 'deepika', 'Hasika','Hasika'],
    'Age': [25,30,30,35,40,40]
})
#print original dataset
print(df)
df_extract=df.drop_duplicates()
print(df_extract)
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    'Id': [1,2,2,3,4,4],
    'Name': ['sindhu', 'rishitha', 'rishitha', 'hasini', 'pranitha','pranitha'],
    'Age': [25,30,40,35,40,40]
})
#print original dataset
print(df)
#remove duplicates on id
df_subset_id=df.drop_duplicates(subset=['Id'])
print(df_subset_id)
#remove duplicaes on name
df_subset_name=df.drop_duplicates(subset=['Name'])
print(df_subset_name)
```

```python
import pandas as pd
df=pd.DataFrame({
    'Data':['2025-91-05','03-03-2023','jan 4,2025','2025.01.05']
})
print("Original Data:\n",df)

#COVERT ALL DATES TO STANDARD ISO FORMAT (YYYY-MM-DD)
df['Data'] = pd.to_datetime(df['Data'], errors='coerce').dt.strftime('%Y-%m-%d')
print(df)
```

```python
#Case Normalization (lowercase/uppercase)
#python
#Sample dataset with inconsistant text cases
df = pd.DataFrame({
    'Name': ['alice', 'bob', 'neo', 'david']})

#convert ali names to lowercase
df['Name_lower'] = df['Name'].str.lower()
#convert ali names to uppercase
df['Name_upper'] = df['Name'].str.upper()
```
